```python
import pandas as pd

data = {
    "Hours": [2.5, 5.1, 3.2, 8.5, 3.5, 1.5, 9.2, 5.5, 8.3, 2.7,
              7.7, 5.9, 4.5, 3.3, 1.1, 8.9, 2.5, 1.9, 6.1, 7.4],
    "Scores": [21, 47, 27, 75, 30, 20, 88, 60, 81, 25,
               85, 62, 41, 42, 17, 95, 30, 24, 67, 69]
}
df = pd.DataFrame(data)

print(df.head())
print(df.shape)
```

```python
print(df.describe())
print("---")
print("Missing values:")
print(df.isnull().sum())
```

```python
# Insight 1 - average score
print("Average score:", df["Scores"].mean())

# Insight 2 - how many scored above 70
high_scorers = df[df["Scores"] > 70]
print("Students who scored above 70:", len(high_scorers))

# Insight 3 - correlation
correlation = df["Hours"].corr(df["Scores"])
print("Correlation between hours and scores:", round(correlation, 2))
```

```python
import matplotlib.pyplot as plt

# Chart 1 - scatter plot
plt.figure(figsize=(7, 4))
plt.scatter(df["Hours"], df["Scores"], color="steelblue", s=80)
plt.xlabel("Hours Studied")
plt.ylabel("Score")
plt.title("Hours Studied vs Exam Score")
plt.tight_layout()
plt.savefig("chart1.png")
plt.show()

# Chart 2 - histogram
plt.figure(figsize=(7, 4))
plt.hist(df["Scores"], bins=6, color="mediumseagreen", edgecolor="white")
plt.xlabel("Score")
plt.ylabel("Number of Students")
plt.title("Distribution of Exam Scores")
plt.tight_layout()
plt.savefig("chart2.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")

# Chart 1 - scatter with trend line
import numpy as np

plt.figure(figsize=(7, 4))
plt.scatter(df["Hours"], df["Scores"], color="#2E86AB", s=80, zorder=3)

# adds a trend line
m, b = np.polyfit(df["Hours"], df["Scores"], 1)
plt.plot(df["Hours"], m * df["Hours"] + b, color="#E84855", linewidth=2)

plt.xlabel("Hours Studied", fontsize=11)
plt.ylabel("Exam Score", fontsize=11)
plt.title("More Study Hours = Higher Scores (r = 0.98)", fontsize=13)
plt.tight_layout()
plt.savefig("chart1_final.png", dpi=150)
plt.show()

# Chart 2 - cleaner histogram
plt.figure(figsize=(7, 4))
plt.hist(df["Scores"], bins=6, color="#2E86AB", edgecolor="white", linewidth=1.2)
plt.axvline(df["Scores"].mean(), color="#E84855", linewidth=2,
            linestyle="--", label=f'Average: {df["Scores"].mean()}')
plt.xlabel("Exam Score", fontsize=11)
plt.ylabel("Number of Students", fontsize=11)
plt.title("Distribution of Exam Scores", fontsize=13)
plt.legend()
plt.tight_layout()
plt.savefig("chart2_final.png", dpi=150)
plt.show()
```
